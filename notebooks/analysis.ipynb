# Study Time and Academic Performance

This notebook performs the preregistered analysis.

```python
import pandas as pd
from scipy.stats import pearsonr, spearmanr

df = pd.read_csv('../data/raw/students.csv')
df = df.drop_duplicates()
df = df.dropna(subset=['study_time', 'final_score'])

print('Sample size:', len(df))
```

```python
r, p = pearsonr(df['study_time'], df['final_score'])
print('Pearson correlation:', r)
print('p-value:', p)
```
